```python
!pip install inference_sdk
import gradio as gr
from PIL import Image, ImageDraw
from inference_sdk import InferenceHTTPClient

CLIENT = InferenceHTTPClient(
    api_url="https://serverless.roboflow.com",
    api_key="ShCHOmtBtUR5CZvqj0dp"
)

MODEL_ID = "waste-detection-pyi5b/1"  #model ID from roboflow
```

```python
def detect(image):

    image.save("temp.jpg")

    result = CLIENT.infer(
        "temp.jpg",
        model_id=MODEL_ID
    )

    img = Image.open("temp.jpg")
    draw = ImageDraw.Draw(img)

    for pred in result['predictions']:

        x = pred['x']
        y = pred['y']
        w = pred['width']
        h = pred['height']

        draw.rectangle(
            [
                (x-w/2, y-h/2),
                (x+w/2, y+h/2)
            ],
            outline="red",
            width=3
        )

        draw.text(
            (x-w/2, y-h/2-15),
            pred['class'],
            fill="red"
        )

    return img
```

```python
demo = gr.Interface(
    fn=detect,
    inputs=gr.Image(type="pil"),
    outputs=gr.Image(),
    title="waste Detection"
)

demo.launch(share=True)
```
